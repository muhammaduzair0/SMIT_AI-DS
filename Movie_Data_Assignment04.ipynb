### **Question 38**

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
sns.barplot(x='Series_Title', y='Score', hue='Metric', data=data.nlargest(5, 'IMDB_Rating')[['Series_Title', 'IMDB_Rating', 'Meta_score']].assign(IMDB_Rating_Scaled=lambda d: d['IMDB_Rating']*10).melt(id_vars='Series_Title', value_vars=['Meta_score', 'IMDB_Rating_Scaled'], var_name='Metric', value_name='Score')).set_xticklabels(plt.gca().get_xticklabels(), rotation=30, ha='right')
```
