```python
def detecting_outliers(data):
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    return lb, ub
```

```python
detecting_outliers(hotel_data['lead_time'])[0]
detecting_outliers(hotel_data['lead_time'])[1]
```
